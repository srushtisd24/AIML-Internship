```python
q1=data.quantile(0.25)
q3=data.quantile(0.75) 
iqr=q3-q1 
lower_limit=q1-1.5*iqr 
upper_limit=q3+1.5*iqr 
print(lower_limit) 
print(upper_limit)
```

```python
outliers = ((data < lower_limit) | (data > upper_limit)).any(axis = 1)
data[outliers]
```

```python
outliers = ((data < lower_limit) | (data > upper_limit)).any(axis = 1)
outlier_data = data[outliers]
Cleaned_data = data[-outliers]
Cleaned_data.head()
```

```python
fig,axs = plt.subplots(1,4,figsize=(4,5))
sns.boxplot(Cleaned_data.Age)
sns.boxplot(Cleaned_data.Salary)
sns.boxplot(Cleaned_data.Experience)
sns.boxplot(Cleaned_data.Monthly_Sales)
plt.show()

```
