```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import linkage,dendrogram
from sklearn.preprocessing import StandardScaler
```

```python
data=pd.read_csv("C:\\Users\\srush\\OneDrive\\Desktop\\AIML Internship\\AIML-Internship\\Day32_Hirarical_clustering\\Mall_Customers.csv")
data.head
```

```python
x = data[["Annual Income (k$)","Spending Score (1-100)"]]
```

```python
scaler= StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
plt.figure(figsize=(15,6))
Z = linkage(x_scaled,method="ward")
dendrogram(Z)
plt.title("dendrogram for Mall Customers")
plt.xlabel("customers")
plt.ylabel("Euclidean Distance")

plt.show()
```

```python
model = AgglomerativeClustering(n_clusters=5)
labels= model.fit_predict(x_scaled)
```

```python
labels
```

```python
data["labels"] = labels
```

```python
## Visuvalizing the Data
plt.scatter(x["Annual Income (k$)"],x["Spending Score (1-100)"],c=data.labels)
plt.xlabel("Annual Income ")
plt.ylabel("Spending Score ")
plt.show()
```
